```python
# Pick any SRR that exists in your data folder
srr = "SRR960281"
infile = f"Flu_Ig_complete_repertoires/recon_input_{srr}_hIgG_annotated.txt"

X, counts = zip(*[line.split() for line in open(infile)])
X = np.array(X)
counts = pd.DataFrame({srr: [int(i) for i in counts]})

# Test the two calls
D0 = relative_sentropy(counts, viewpoint=[0], measures=['alpha'], parallelize=True)['set_alpha_q=0']
D0S = relative_sentropy(
        counts,
        similarity=lambda a,b: 0.3**distance(a,b),
        X=X,
        viewpoint=[0],
        measures=['alpha'],
        parallelize=True,
        symmetric=True,
    )['set_alpha_q=0']

print("D0 =", D0, "D0S =", D0S)
```

```python
def make_S(seqs):
    S = np.identity(len(seqs))
    for i, j in combinations(range(len(seqs)), 2):
        s_ij = 0.3**distance(seqs[i], seqs[j])       # from Arora 2022 PNAS
        S[i][j] = s_ij
        S[j][i] = s_ij
    return S
```

```python
labels, counts = zip(*[line.split() for line in open(infile)])
counts = [int(i) for i in counts]
counts = pd.DataFrame({srr: counts})

m = Metacommunity(counts)
print(m.subcommunity_diversity(viewpoint=0, measure='alpha')[0])

S = make_S(labels)
mS = Metacommunity(counts, similarity=S)
print(mS.subcommunity_diversity(viewpoint=0, measure='alpha')[0])
```
