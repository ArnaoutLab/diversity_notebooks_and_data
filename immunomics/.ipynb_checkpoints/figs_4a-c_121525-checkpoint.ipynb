(In fact, we expect the corresponding patterns in the overall repertoires to be quantitatively more dramatic, as sampling decreases the differences in diversities among repertoires.)

````python
def subsample(inpath, n, outdir="", verbose=False, debug=False, quiet=False):
    """
    If inpath is a file, samples from that file. If inpath is a directory, assumes all files in that directory belong to the same sample, reads them all into a single data structure, and then subsamples from that.
    n is either the number of sequences to subsample (i.e., the subsample size), or a list of such sizes.
    If outdir not specified, places output in the same directory as inpath (if inpath is a file, then in the same directory; if inpath is already a directory, then in inpath)
    """
    #
    # Read in data
    if os.path.isdir(inpath):
        if not inpath.endswith("/"): inpath += "/"
        list_of_files = glob(inpath + "*")
    else: 
        list_of_files = [inpath]
    if verbose:
        now = datetime.now()
        print(now, "\tgetting sequences and their frequencies...")
        sys.stdout.flush()
    pooled_seqs = defaultdict(float)
    for f in list_of_files:
        if verbose:
            print("%s\t\t%s" % (datetime.now(), f))
            sys.stdout.flush()
        try:
            for (cdr1, cdr2, cdr3), count_ in parse_file_contents(f, quantitative=True):
                pooled_seqs[cdr3] += count_
        except: # recon-formatted file (CDRE tab integer count)
            for line in open(f):
                cdr3, count_ = line.rstrip().split()
                count_ = int(count_)
                pooled_seqs[cdr3] += count_
        if debug: break
    unique_pooled_seqs, frequencies = zip(*pooled_seqs.items())
    sum_frequencies = sum(frequencies)
    frequencies = [float(i)/sum_frequencies for i in frequencies]
    assert len(unique_pooled_seqs) == len(frequencies)
    #
    # downsample and write to file
    if type(n) in (int, float): 
        n = [n]
    now = datetime.now()
    if verbose: 
        print(now, "\tsampling and writing...")
    if outdir == "":
        outdir = inpath.rsplit("/", 1)[0]
    for size in n:
        size = int(size)
        outfile = inpath.rstrip("/").rsplit("/", 1)[1].rsplit(".", 1)[0]
        random_code = ''.join( [np.random.choice( list(string.ascii_uppercase + string.ascii_lowercase + string.digits) ) for _ in range(5)] )
        suffix = "_%.0e_%s.txt" % (size, random_code)
        suffix = suffix.replace("+0", "")
        outfile = outdir + outfile + suffix
        if not quiet:
            print( "%s\t\t%s"% (datetime.now(), outfile) )
        size = int(size)
        sample_list = np.random.choice(unique_pooled_seqs, size, p=frequencies)
        c = Counter(sample_list)
        with open(outfile, 'w') as f:
            for k, v in c.items():
                val_str = "%s\t%s\n" % (k,v)
                f.write(val_str)
    return
````

````python
# The below is adapted from build_runid_size_q_hash
# files_to_compare = glob(flu_repertoire_subsamples_for_figs_4ab_dir + "*")
files_to_compare = flu_subsample_files
filename_must_start_with = "recon_input_SRR"
debug = True

sample_ids = set()
sample_sizes = set()
completed_run_ids = set()
completed_repertoires = set()
allowed_repertoires = set([i.split(".")[0].rsplit("/", 1)[-1] for i in files_to_compare])                 # remove the path and suffix
allowed_repertoires = set([i for i in allowed_repertoires if i.startswith(filename_must_start_with)])
reobj = re.compile("_([\d]e[\d])[_.]")                                                                    # for extracting sample size from filenames

# print("allowed_repertoires:")
# for i in sorted(list(allowed_repertoires)):
#     print(i)
# print()

H = dict()
#
#earliest_date = "2020-12-07 09:00:00"                                               # this was when we reran using full repertoires for Recon qD_est ("qD" below)
                                            # this was when we rechecked this for shifting from morty to sentropy
earliest_date = "2025-11-01 00:00:00"
earliest_date = datetime.strptime(earliest_date, "%Y-%m-%d %H:%M:%S")
for line in open(alpha_diversity_master_file): 
    try:
        run_id, measure, repertoire, q_hash, date, command = line.strip().split("\t")        
    except:
        run_id, measure, repertoire, qDs_hash, q_hash_obsD, qD_hash, date, command = line.strip().split("\t")
        qDs_hash    = ast.literal_eval(qDs_hash)
        q_hash_obsD = ast.literal_eval(q_hash_obsD)
        qD_hash     = qD_hash.replace(": inf", ": 'inf'")
        qD_hash     = ast.literal_eval(qD_hash)    
        q_hash      = {**qDs_hash, **q_hash_obsD, **qD_hash}
    
    # filter
    if date == "date": continue              # the header line contains "date"
    date = datetime.strptime(date, "%Y-%m-%d %H:%M:%S")
    if date <= earliest_date: continue
    if repertoire in completed_run_ids: continue
    if repertoire in completed_repertoires: continue
    if repertoire not in allowed_repertoires: continue
    # print(date, repertoire)
    completed_run_ids.add(run_id)
    completed_repertoires.add(repertoire)
    try: full_path_infile = command.split("-if")[1].split("-mod")[0].strip()      # this command structure was used for the runs we desire
    except: 
        continue
    
    #
    # catch cases where the file was run as another user
    if expanduser("~") not in full_path_infile:
        items = full_path_infile.split("/")
        user_item = items.index("Users") + 1                       # assumes /Users/<username>/rest/of/path
        items = items[user_item+1:]
        full_path_infile = expanduser("~") + "/" + "/".join(items)
    #
    # process
    infile = full_path_infile.rsplit("/", 1)[-1]
    try:
        sample_size = reobj.search(infile).groups()[0]
    except:           # the full-repertoire file (as opposed to subsample files) has no sample size in the filename; this catches 
        if "50K" in infile: sample_size = "5e4"
        else: sample_size = False
    sample_size = float(sample_size)
    try:
        qDs_hash, qD_hash = eval(q_hash)
    except: 
        pass
    qDs_hash.update(qD_hash)
    for k, v in qDs_hash.items():                        # convert to floats
        qDs_hash[k] = float(v)
    #
    k = repertoire.split("_")[2]                         # SRR-specific line
    H[k] = qDs_hash
````

````python
# get values (pair for swarmplot, but be sure to uniquify before getting summary statistics---i.e., don't double-count)
def get_qD_values(d0_srrs, d7_srrs, type_of_diversity, q="0.0", debug=False):
    pairs = []
    if debug: print(T_ids_with_d0_and_d7)
    for _id in T_ids_with_d0_and_d7:
        d0_srrs = [i[0] for i in id_d0_igg_T_srr_hash[_id]]
        d7_srrs = [i[0] for i in id_d7_igg_T_srr_hash[_id]]
        for d0_srr, d7_srr in product(d0_srrs, d7_srrs):
            if type_of_diversity == "sequence":
                if "%sD" % q in H[d0_srr]:
                    d0_srr_val = H[d0_srr]["%sD" % q]
                elif "est_%sD" % q in H[d0_srr]:
                    d0_srr_val = H[d0_srr]["est_%sD" % q]
                if "%sD" % q in H[d7_srr]:
                    d7_srr_val = H[d7_srr]["%sD" % q]
                elif "est_%sD" % q in H[d7_srr]:
                    d7_srr_val = H[d7_srr]["est_%sD" % q]
            elif type_of_diversity == "class":
                diversity_key = "%sDs" % q
                d0_srr_val = H[d0_srr][diversity_key]
                d7_srr_val = H[d7_srr][diversity_key]
            pairs.append((d0_srr_val, d7_srr_val))
    day0_values, day7_values = zip(*pairs)
    return day0_values, day7_values

day0_TIV_vaccine_SRR_0D,  day7_TIV_vaccine_SRR_0D  = get_qD_values(d0_igg_TIV_srrs, d7_igg_TIV_srrs, "sequence")
day0_TIV_vaccine_SRR_0Ds, day7_TIV_vaccine_SRR_0Ds = get_qD_values(d0_igg_TIV_srrs, d7_igg_TIV_srrs, "class")
````

````python
def round_to_nearest(x, to_nearest=10):
    return int(round(x / to_nearest)) * to_nearest

def make_swarmplot(df, palette=None, figsize=(2,3), rotation=90, plot_all=True, size=6, save=False, filename="", debug=False, replicates=()):
    """
    df must have two columns. This plots a swarmplot of the first column, and next to that a swarmplot of the second column, and on top of that a line for the means.

    replicates is a list of row_indices that indicates any values that are replicates of the row immediately after
    """
    if debug:
        print(df)
    #
    # calculate median
    first_col =  np.array(df.iloc[:,0]).astype(int)
    second_col = np.array(df.iloc[:,1]).astype(int)
    # handle replicates
    first_col_for_median = []
    second_col_for_median = []
    for i in range(len(first_col)):
        if i in replicates:
            first_col_for_median.append(  (first_col[i]  + first_col[i+1])  / 2 )
            second_col_for_median.append( (second_col[i] + second_col[i+1]) / 2 )
            i += 1 # skips the next line
        else:
            first_col_for_median.append( first_col[i] )
            second_col_for_median.append( second_col[i] )        
    first_col_median = np.median(first_col_for_median)
    second_col_median = np.median(second_col_for_median)
    pct_diff_of_medians = (second_col_median/first_col_median - 1) * 100
    #
    # calculate median change
    pct_diffs = ( np.array( df.iloc[:,1] / df.iloc[:,0] ) - 1. ) * 100
    pct_median_of_diffs = np.median(pct_diffs)
    print("Medians: %.0f -> %.0f (%+.0f%%)" % (first_col_median, second_col_median, pct_diff_of_medians) )
    print("pct_diffs:", sorted([int(i) for i in pct_diffs]))   # int just for presentation purposes
    print("Median of changes from d0 to d7: %+.0f%%" % pct_median_of_diffs)
    #
    # make plot
    plt.ion()
    plt.figure(figsize=figsize)
    #
    # plot swarmplot
    df = df.melt()      # formats for sns.swarmplot
    g = sns.swarmplot(x="variable", y="value", data=df, edgecolor="black", linewidth=1, palette=palette, size=size)
    #
    # plot trend
    if plot_all:
        # lookup jittered x-values
        x, y = np.array(g.collections[0].get_offsets()).T
        y = y.astype(int)
        if debug:
            print("x:", x)
            print("y:", y)
        x_lookup_1 = { round_to_nearest(y_i, to_nearest=10):x_i for x_i, y_i in zip(x, y) }
        if debug:
            print("x_lookup_1:", x_lookup_1)
            
        x, y = np.array(g.collections[1].get_offsets()).T
        x_lookup_2 = { round_to_nearest(y_i, to_nearest=10):x_i for x_i, y_i in zip(x, y) }
        for first_val, second_val in zip(first_col, second_col):
            first_x = x_lookup_1[round_to_nearest(first_val, to_nearest=10)]
            second_x = x_lookup_2[round_to_nearest(second_val, to_nearest=10)]
            plt.plot([first_x, second_x], [first_val, second_val], c=colors["red"], alpha=0.2, linewidth=1)
    plt.plot([0, 1], [first_col_median, second_col_median], c=colors["red"], linewidth=2)                          # plot summary
    #
    # formatting
    g.set_xticklabels(g.get_xticklabels(), rotation=rotation)
    plt.yscale('log')
    #
    # save
    if save:
        plt.savefig(filename, transparent=True)
    #
    # show
    # plt.tight_layout()
    plt.show()
    return
````

Functions:

````python
def subsample_from_all_files_in_directory(repertoire_indir, subsample_outdir):
    """
    This harness runs subsample() on each of the files in repertoire_indir and puts them in subsample_outdir.
    It then runs a check just to make sure subsample() worked.
    repertoire_indir   contains the complete repertoires that are to be subsampled
    subsample_outdir   where the subsamples are deposited
    """
    # create subsamples
    for recon_input_file in repertoire_indir:
        subsample(recon_input_file, sample_size, outdir=subsample_outdir, quiet=True)
    #
    # check that subsampling gave files of the right size
    subsample_files = glob(subsample_outdir + "*")
    subsample_files = [i for i in subsample_files if not isdir(i)] # in case there's also a directory in this folder (e.g. for a previous set of subsamples)
    for i in subsample_files:
        tot = 0
        for line in open(i):
            cdr3, no = line.strip().split()
            tot += int(no)
        if not tot == sample_size: print("error: %s: %i != %i" % (i, tot, sample_size))
    return
````

````python
def generate_qD_vs_qDs_pairs(allowed_repertoires, subsample_size=1e5, q='0.0', debug=False, alpha_diversity_master_file=alpha_diversity_master_file, earliest_date="2020-12-07 09:00:00"):
    """
    earliest_date is when we reran using full repertoires for Recon qD_est (referenced as "qD" in the actual file)
    """
    reobj = re.compile("_([\d]e[\d])[_.]")                                                     # for extracting sample size from filenames
    completed_run_ids = set()
    completed_repertoires = set()
    pairs_list = []
    earliest_date = datetime.strptime(earliest_date, "%Y-%m-%d %H:%M:%S")
    for line_no, line in enumerate(open(alpha_diversity_master_file), 1):
        run_id, measure, repertoire, q_hash, date, command = line.strip().split("\t")
        if date == "date": continue                                                            # the header line contains "date", and it's somewhere in the file
        date = datetime.strptime(date, "%Y-%m-%d %H:%M:%S")
        if date <= earliest_date:
            if debug:
                print(line_no, "skipping: date <= earliest_date:", date, earliest_date)
            continue
        if debug:
            print(line_no, "\t", repertoire)
        if debug and repertoire in allowed_repertoires:
            print("found\t", line)
        if run_id in completed_run_ids \
            or repertoire in completed_repertoires \
            or repertoire not in allowed_repertoires:
            continue
        if debug:
            print(run_id, repertoire)
        completed_run_ids.add(run_id)
        completed_repertoires.add(repertoire)
        try: full_path_infile = command.split("-if")[1].split("-mod")[0].strip()              # this command structure was used for the runs we desire
        except: continue
        if debug:
            print(full_path_infile)
        # catch cases where the file was run as another user
        if expanduser("~") not in full_path_infile:
            items = full_path_infile.split("/")
            user_item = items.index("Users") + 1                                              # assumes /Users/<username>/rest/of/path
            items = items[user_item+1:]
            full_path_infile = expanduser("~") + "/" + "/".join(items)
            # print("full_path_infile:", full_path_infile)        
        infile = full_path_infile.rsplit("/", 1)[-1]
        if debug:
            print(infile)
        try:
            sample_size = reobj.search(infile).groups()[0]
        except:                                              # the full-repertoire file (as opposed to subsample files) has no sample size in the filename; this catches 
            if "50K" in infile: sample_size = 5e4
            else: sample_size = False
        if debug:
            print(sample_size)
        sample_size = float(sample_size)
        if sample_size != subsample_size:
            continue
        qDs_hash, qD_hash = eval(q_hash)
        qD_value  = qD_hash['%sD' % q]
        qDs_value = qDs_hash['%sDs' % q]
        pair = (qD_value, qDs_value)
        pairs_list.append(pair)
    return pairs_list
````

We load 1e6 subsamples from the subsamples we took there, not the usual 1e5, because 1e6 is possible, and they will give a more accurate R2 than the 1e5s we are limited to with the mRNA repertoires we have.

````python
# relative_ordering_test_dir = expanduser("~") + "/Dropbox (ArnaoutLab)/Harry/repo/Python scripts/Alpha_diversity_paper/Alpha_paper_revisions/Figures/Fig3/Relative_ordering_test/"
relative_ordering_test_dir = expanduser("~") + "/Documents/Papers_2022/Alpha_diversity_paper/Old/Alpha_paper_revisions/Figures/Fig3/Relative_ordering_test/"
igh_subsample_dirs = [i for i in glob(relative_ordering_test_dir + "/*") if i.rsplit("/", 1)[-1].startswith("D") and i.endswith("_subsamples")]  # Directories are D1, D2, etc.
sample_names = [i.rsplit("/", 1)[-1].split("_")[0] for i in igh_subsample_dirs]

sample_size = '1e6'
sample_name_subsample_list_hash = {}
for igh_subsample_dir in igh_subsample_dirs:
    sample_name = igh_subsample_dir.rsplit("/", 1)[-1].split("_")[0]            # same as above
    subsample_files = glob(igh_subsample_dir + "/*")
    subsample_files = [i for i in subsample_files if sample_size in i]
    subsample_file = subsample_files[0]                                         # there's only one 1e6
    sample_name_subsample_list_hash[sample_name] = subsample_file
subsample_files = list(sample_name_subsample_list_hash.values())

# revisions_folder = expanduser("~") + "/Dropbox (ArnaoutLab)/Harry/repo/Python scripts/Alpha_diversity_paper/Alpha_paper_revisions/"
revisions_folder = expanduser("~") + "/Documents/Papers_2022/Alpha_diversity_paper/Old/Alpha_paper_revisions/"
alpha_diversity_master_file = revisions_folder + "alpha_diversity_master_file.txt"

allowed_repertoires = [subsample_file.rsplit("/", 1)[-1].rsplit(".", 1)[0] for subsample_file in subsample_files]
igh_dna_fig4c_0D_0Ds_pairs_list = generate_qD_vs_qDs_pairs(allowed_repertoires, subsample_size=1e6, earliest_date="2020-11-10 00:00:00")

igh_dna_0D, igh_dna_0Ds = list(zip(*igh_dna_fig4c_0D_0Ds_pairs_list))
igh_dna_0D  = np.array(igh_dna_0D).astype(float)
igh_dna_0Ds = np.array(igh_dna_0Ds).astype(float)
igh_dna_regr = linregress(igh_dna_0D, igh_dna_0Ds)
igh_dna_r2 = igh_dna_regr.rvalue**2
no_igh_dna = len(igh_dna_0D)
print("igh_dna_r2 (n=%i): %.2f" % (no_igh_dna, igh_dna_r2))
````
